```python
import pandas as pd 
import numpy as np 
import pickle

with open("master_database.pkl", "rb") as f:
    df = pickle.load(f)

df
```

```python
presentation_charts = {
    "model_version": "v1.0.0",
    "customer_id": "",
    "currency_id": "",
    "forecasts": []
}

counter = 0
for i in df["customer_id"].unique():
    if counter > 3 : break 
    
    highlight_df = df[df["customer_id"] == i]
    highlight_df["room_type_id"] = highlight_df["room_type_id"].astype(int)
    highlight_df = highlight_df.sort_values("room_type_id", ascending=True)

    id_storage = i
    curr_id_storage = highlight_df["currency_id"].iloc[0]

    highlight_df = highlight_df.drop(columns=["customer_id", "currency_id", "flag", "_source"])
    presentation_charts_copy = presentation_charts.copy()
    interest_json = highlight_df.to_dict(orient="records")
    presentation_charts_copy["customer_id"] = id_storage
    presentation_charts_copy["currency_id"] = curr_id_storage
    sep_box = []
    inner_counter = 0
    for ii in range(len(interest_json)):
        if inner_counter > 3: break
        print(interest_json[ii])
        sep_box.append(interest_json[ii])
        inner_counter += 1
    presentation_charts_copy["forecasts"] = sep_box
    print(presentation_charts_copy["forecasts"])
    counter += 1
```
